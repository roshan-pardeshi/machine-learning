```python
df = pd.read_csv(r"C:\Users\Admin\Downloads\maharashtra_house_prices_missing_values_15000.csv")
```

```python
df["city"] = df["city"].fillna(df["city"].mode()[0])
df["property_type"] = df["property_type"].fillna(df["property_type"].mode()[0])
df["area_sqft"] = df["area_sqft"].fillna(df["area_sqft"].mean())              
df["bedrooms"] = df["bedrooms"].fillna(df["bedrooms"].mean())                 
df["bathrooms"] = df["bathrooms"].fillna(df["bathrooms"].mean()) 
df["property_age_years"] = df["property_age_years"].fillna(df["property_age_years"].mean())     
df["parking_spaces"] = df["parking_spaces"].fillna(df["parking_spaces"].mean())          
df["furnishing"] = df["furnishing"].fillna(df["furnishing"].mode()[0])             
df["distance_to_city_km"] = df["distance_to_city_km"].fillna(df["distance_to_city_km"].mean())    
df["nearby_schools_count"]=df["nearby_schools_count"].fillna(df["nearby_schools_count"].median())    
df["crime_index"] =  df["crime_index"].fillna(df["crime_index"].mean())           
df["construction_status"] = df["construction_status"].fillna(df["construction_status"].mode()[0])      
df["sale_price_inr"] = df["sale_price_inr"].fillna(df["sale_price_inr"].mean())           
```

```python
df.isna().sum()
```

```python
le = LabelEncoder()

df["city"] = le.fit_transform(df["city"])
df["property_type"] = le.fit_transform(df["property_type"])

df["furnishing"] = le.fit_transform(df["furnishing"])

df["construction_status"] = le.fit_transform(df["construction_status"])
```

```python
df.drop(columns="property_id",inplace=True)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error,r2_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score
```

```python
model = LinearRegression()

x = df[["city","property_type","area_sqft","bedrooms","bathrooms","property_age_years","parking_spaces","furnishing","distance_to_city_km","nearby_schools_count","crime_index","construction_status"]]
y = df["sale_price_inr"]

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=222)
```

```python
model.fit(x_train,y_train)

y_predict = model.predict(x_test)
```

```python
mae = mean_absolute_error(y_test,y_predict)
```
