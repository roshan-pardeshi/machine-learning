```python
def arm(n):

    num = n
    cube = 0
    num1 = len(str(num))
    no = num
    for i in range(0,num,1):
        a=no%10
        cube = cube+a**num1
        no=no//10
    
    if cube==num:
        print("armstrong number",num)
    

for i in range(1,1000,1):
    arm(i)
```
