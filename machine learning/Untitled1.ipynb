```python
def prime(n):
    num = n
    factor = 0

    for i in range(2,num-1,1):
        if num%i==0:
            factor+=1

    if factor==0:
        print("Prime Number",num)

for i in range(1,101,1):
    prime(i)
```

```python
def uniques(n):
    li = []
    
    for i in n:
        if i not in li:
            li.append(i)

    return li

n = [10,20,30,40,40,50]

for i in range(len(n)):
    uniques(i)
```
