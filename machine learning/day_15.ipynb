```python
num = 4

factor = 0

for i in range(2,num):
    if num%i==0:
        factor+=1
if factor==0:
    print("is an prime number")
else:
    print("Not a Prime")
```

```python
def prime(n):
    num=n
    factor=0
    for i in range(2,num):
        if num%i==0:
            factor+=1

    if factor==0:
        print("Prime",num)

for i in range(1,101,1):
    prime(i)
```
