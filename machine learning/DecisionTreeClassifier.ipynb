```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv(r"C:\Users\Admin\Downloads\loan_default_practice_5000.csv")
```

```python
df.head()
```

```python
df.duplicated().sum()
```

```python
df.isna().sum()
```

```python
df["Education"] = df["Education"].fillna(df["Education"].mode()[0])
```

```python
df.isna().sum()
```

```python
df["EmploymentStatus"] = df["EmploymentStatus"].fillna(df["EmploymentStatus"].mode()[0])
```

```python
df["MonthlyIncome"] = df["MonthlyIncome"].fillna(df["MonthlyIncome"].mean())
```

```python
df["CreditScore"] = df["CreditScore"].fillna(df["CreditScore"].median())
```

```python
df["DebtToIncomeRatio"] = df["DebtToIncomeRatio"].fillna(df["DebtToIncomeRatio"].mean())
```

```python
df.dtypes
```

```python
df.select_dtypes('object')
```

```python
df=df.iloc[0::1,1::1]
```

```python
df.select_dtypes('object')
```

```python
# Gender	Education	MaritalStatus	EmploymentStatus	HomeOwnership	LoanPurpose	Defaulted
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
df["Gender"] = le.fit_transform(df["Gender"])
df["Education"] = le.fit_transform(df["Education"])
df["MaritalStatus"]= le.fit_transform(df["MaritalStatus"])
df["EmploymentStatus"] = le.fit_transform(df["EmploymentStatus"])
```

```python
df["HomeOwnership"] = le.fit_transform(df["HomeOwnership"])
```

```python
df["LoanPurpose"] = le.fit_transform(df["LoanPurpose"])
```

```python
df["Defaulted"] = le.fit_transform(df["Defaulted"])
```

```python
x = df[["Age","Gender","Education","MaritalStatus","EmploymentStatus","HomeOwnership","LoanPurpose","MonthlyIncome","CreditScore","LoanAmount","LoanTermMonths","ExistingLoans","PreviousDefaults","EmploymentYears","DebtToIncomeRatio"]]
```

```python
y = df["Defaulted"]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=22)
```

```python
model = DecisionTreeClassifier()
```

```python
model.fit(x_train,y_train)
```

```python
y_predict = model.predict(x_test)
```

```python
y_predict[0:10]
```

```python
from sklearn.metrics import accuracy_score
```

```python
ac = accuracy_score(y_test,y_predict)
```
