```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error,r2_score
```

```python
df = pd.read_csv(r"C:\Users\Admin\Downloads\house_price_ml_practice_dirty.csv")
```

```python
df.head()
```

```python
df[df["Area_sqft"]<0]
```

```python
df["Area_sqft"] = df["Area_sqft"].apply(lambda x:x*(-1) if x<0 else x*1)
```

```python
df.isnull().sum()
```

```python
df["Area_sqft"] = df["Area_sqft"].fillna(df["Area_sqft"].mean())
df["Bedrooms"] = df["Bedrooms"].fillna(df["Bedrooms"].median())
df["Bathrooms"] = df["Bathrooms"].fillna(df["Bathrooms"].mean())
df["Age_years"] = df["Age_years"].fillna(df["Age_years"].mean())
df["Distance_to_City_km"] = df["Distance_to_City_km"].fillna(df["Distance_to_City_km"].mean())
a = df["Location"].mode()
df["Location"] =df["Location"].fillna(a[0])
b = df["Parking"].mode()
df["Parking"] = df["Parking"].fillna(b[0])
c = df["Furnishing"].mode()
df["Furnishing"] = df["Furnishing"].fillna(c[0])
df["Price_INR"] = df["Price_INR"].fillna(df["Price_INR"].mean())
```

```python
df.isna().sum()
```

```python
df.groupby("Property_Type")["Price_INR"].mean()
```

```python
df["Property_Type"] = df["Property_Type"].str.strip()
```

```python
df["Property_Type"] = df["Property_Type"].apply(lambda x:x.split()[0])
```

```python
df["Property_Type"].apply(lambda x: x.split()[0] if pd.notna(x) else x)
```

```python
df.select_dtypes("float64")
```

```python
df.columns
```

```python
x = df[['Area_sqft', 'Bedrooms', 'Bathrooms', 'Age_years',
       'Distance_to_City_km']]
```

```python
y = df['Price_INR']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=234)
```

```python
model = LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
res = model.predict(x_test)
```
