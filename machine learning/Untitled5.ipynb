```python
df.head()

df["area_sqft"] = df["area_sqft"].astype("Int64")

df["bedrooms"] = df["bedrooms"].astype("Int64")

df["bathrooms"] = df["bathrooms"].astype("Int64")

df["property_age_years"] = df["property_age_years"].astype("Int64")

df["parking_spaces"] = df["parking_spaces"].astype("Int64")

# df["distance_to_city_km"] = df["distance_to_city_km"].astype("Int64")

df["nearby_schools_count"] = df["nearby_schools_count"].astype("Int64")

df["sale_price_inr"] = df["sale_price_inr"].astype("Int64")
```

```python
df["city"] = df["city"].fillna(df["city"].mode()[0])

df["property_type"] = df["property_type"].fillna(df["property_type"].mode()[0])

df["area_sqft"] = df["area_sqft"].fillna(round(df["area_sqft"].mean()))

df["bedrooms"] = df["bedrooms"].fillna(round(df["bedrooms"].median()))

df["bathrooms"] = df["bathrooms"].fillna(round(df["bathrooms"].mean()))

df["property_age_years"] = df["property_age_years"].fillna(round(df["property_age_years"].mean()))

df["parking_spaces"] = df["parking_spaces"].fillna(round(df["parking_spaces"].mean()))

df["furnishing"] = df["furnishing"].fillna(df["furnishing"].mode()[0])

df["distance_to_city_km"] = df["distance_to_city_km"].fillna(round(df["distance_to_city_km"].mean()))

df["nearby_schools_count"] = df["nearby_schools_count"].fillna(round(df["nearby_schools_count"].mean()))

df["crime_index"] = df["crime_index"].fillna(round(df["crime_index"].median()))

df["construction_status"] = df["construction_status"].fillna(df["construction_status"].mode()[0])

df["sale_price_inr"] = df["sale_price_inr"].fillna(round(df["sale_price_inr"].mean()))

                                             
```
